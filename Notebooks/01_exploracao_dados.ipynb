## Harmonização e comparação das escolas de 2023 e 2025

O arquivo de 2023 usa `IN_FUND_AF` para indicar oferta de anos finais, enquanto a tabela de 2025 usa `IN_COMUM_FUND_AF`. A leitura abaixo renomeia o campo de 2023 para um nome canônico e mantém apenas escolas públicas em funcionamento de São Paulo que oferecem anos finais. O processamento em blocos limita o uso de memória.

A comparação longitudinal usa somente escolas com `CO_ENTIDADE` presente nos dois anos. Escolas sem correspondência permanecem fora da coorte pareada; não são tratadas como desaparecidas ou novas sem investigação. A equivalência conceitual dos indicadores compartilhados deve ser confirmada no dicionário de cada edição antes de interpretar variações.


```python
caminho_escolas_2023 = DADOS_BRUTOS / "microdados_ed_basica_2023.csv"
caminho_tratado_2023 = DADOS_TRATADOS / "escolas_sp_2023_tratado.csv"

if caminho_tratado_2023.exists():
    # O cache contém a mesma população e colunas já harmonizadas; apague-o para refazer.
    df_escolas_2023 = pd.read_csv(
        caminho_tratado_2023,
        dtype={"CO_ENTIDADE": "string", "CO_MUNICIPIO": "string"},
        encoding="utf-8-sig"
    )
    print("Cache Censo 2023 carregado:", caminho_tratado_2023.relative_to(PASTA_PROJETO))
else:
    # Em 2023, o indicador de oferta dos anos finais se chama IN_FUND_AF.
    colunas_2023 = [
        "IN_FUND_AF" if coluna == "IN_COMUM_FUND_AF" else coluna
        for coluna in colunas_interesse
    ]
    partes_2023 = []

    # Lê o microdado grande em blocos para reduzir o consumo de memória.
    for parte in pd.read_csv(
        caminho_escolas_2023,
        sep=";",
        encoding="latin-1",
        usecols=colunas_2023,
        dtype={"CO_ENTIDADE": "string", "CO_MUNICIPIO": "string"},
        chunksize=100_000,
        low_memory=False
    ):
        # Aplica o recorte comum: rede pública, ativa, em SP e com anos finais.
        filtro_2023 = (
            parte["SG_UF"].eq("SP") &
            parte["TP_DEPENDENCIA"].isin([1, 2, 3]) &
            parte["TP_SITUACAO_FUNCIONAMENTO"].eq(1) &
            parte["IN_FUND_AF"].eq(1)
        )
        partes_2023.append(parte.loc[filtro_2023].copy())

    df_escolas_2023 = pd.concat(partes_2023, ignore_index=True)
    df_escolas_2023 = df_escolas_2023.rename(
        columns={"IN_FUND_AF": "IN_COMUM_FUND_AF"}
    )
    df_escolas_2023["QT_DISPOSITIVOS_ALUNO"] = df_escolas_2023[
        ["QT_DESKTOP_ALUNO", "QT_COMP_PORTATIL_ALUNO", "QT_TABLET_ALUNO"]
    ].sum(axis=1, min_count=1)
    df_escolas_2023.to_csv(
        caminho_tratado_2023,
        index=False,
        encoding="utf-8-sig"
    )
    print("Cache Censo 2023 criado:", caminho_tratado_2023.relative_to(PASTA_PROJETO))

# Padroniza o identificador e prepara os dados de 2025 já tratados acima.
df_escolas_2023["CO_ENTIDADE"] = df_escolas_2023["CO_ENTIDADE"].astype("string")
df_escolas_2023["CO_MUNICIPIO"] = df_escolas_2023["CO_MUNICIPIO"].astype("string")
df_escolas_2023["NU_ANO_CENSO"] = 2023

df_escolas_2025 = df_escolas_tratado.copy()
df_escolas_2025["CO_ENTIDADE"] = df_escolas_2025["CO_ENTIDADE"].astype("string")
df_escolas_2025["CO_MUNICIPIO"] = df_escolas_2025["CO_MUNICIPIO"].astype("string")
df_escolas_2025["NU_ANO_CENSO"] = 2025

# Cada código de escola deve aparecer uma vez por ano; duplicatas invalidariam o merge pareado.
for ano, dados in [(2023, df_escolas_2023), (2025, df_escolas_2025)]:
    print(
        f"{ano}: {len(dados):,} escolas; "
        f"{dados['CO_MUNICIPIO'].nunique():,} municípios; "
        f"IDs ausentes: {dados['CO_ENTIDADE'].isna().sum():,}; "
        f"IDs duplicados: {dados['CO_ENTIDADE'].duplicated().sum():,}"
    )
    if dados["CO_ENTIDADE"].isna().any() or dados["CO_ENTIDADE"].duplicated().any():
        raise ValueError(f"A chave CO_ENTIDADE precisa ser única e não nula em {ano}.")

ids_2023 = set(df_escolas_2023["CO_ENTIDADE"])
ids_2025 = set(df_escolas_2025["CO_ENTIDADE"])
print(f"Escolas na coorte pareada: {len(ids_2023 & ids_2025):,}")
print(f"Escolas somente em 2023: {len(ids_2023 - ids_2025):,}")
print(f"Escolas somente em 2025: {len(ids_2025 - ids_2023):,}")

# O merge é um-para-um: uma linha por escola em cada edição do Censo.
# O inner join cria a coorte de escolas presentes nos dois anos.
df_painel_escolas = df_escolas_2023.merge(
    df_escolas_2025,
    on="CO_ENTIDADE",
    how="inner",
    suffixes=("_2023", "_2025"),
    validate="one_to_one"
)

caminho_painel = DADOS_TRATADOS / "painel_escolas_sp_2023_2025.csv"
df_painel_escolas.to_csv(caminho_painel, index=False, encoding="utf-8-sig")
print("Painel pareado salvo em:", caminho_painel.relative_to(PASTA_PROJETO))
print("Dimensões:", df_painel_escolas.shape)
```

## Inclusão do IDEB por escola

A planilha de IDEB dos anos finais contém resultados observados de 2023 e 2025 por código de escola. O código `ID_ESCOLA` é harmonizado com `CO_ENTIDADE`; a junção é validada como um-para-um e restrita às redes municipal, estadual e federal de São Paulo. Valores como `-` ou `ND` são mantidos como ausentes após a conversão numérica, nunca como zero. A seleção filtrada é armazenada em um CSV de cache e reutilizada enquanto a planilha bruta não for atualizada.


```python
caminho_ideb = DADOS_BRUTOS / "divulgacao_anos_finais_escolas_2025.xlsx"
colunas_ideb = [
    "SG_UF",
    "CO_MUNICIPIO",
    "ID_ESCOLA",
    "REDE",
    "VL_OBSERVADO_2023",
    "VL_OBSERVADO_2025"
]

# Reutiliza o recorte local se ele existir; apague o cache para refazê-lo.
caminho_cache_ideb = (
    DADOS_TRATADOS / "ideb_escolas_sp_publicas_coorte_2023_2025.csv"
)
if caminho_cache_ideb.exists():
    df_ideb = pd.read_csv(
        caminho_cache_ideb,
        dtype={"ID_ESCOLA": "string", "CO_MUNICIPIO": "string"},
        encoding="utf-8-sig"
    )
    print("Cache do IDEB carregado:", caminho_cache_ideb.relative_to(PASTA_PROJETO))
else:
    # Na primeira execução ou após atualização da fonte, filtra e salva o recorte.
    df_ideb = pd.read_excel(
        caminho_ideb,
        sheet_name=0,
        header=9,
        usecols=colunas_ideb,
        dtype={"ID_ESCOLA": "string"},
        engine="openpyxl"
    )
    df_ideb = df_ideb.loc[
        df_ideb["SG_UF"].eq("SP") &
        df_ideb["REDE"].isin(["Municipal", "Estadual", "Federal"])
    ].copy()
    for coluna in ["VL_OBSERVADO_2023", "VL_OBSERVADO_2025"]:
        df_ideb[coluna] = pd.to_numeric(df_ideb[coluna], errors="coerce")
    df_ideb.to_csv(caminho_cache_ideb, index=False, encoding="utf-8-sig")
    print("Cache do IDEB criado:", caminho_cache_ideb.relative_to(PASTA_PROJETO))

# Reaplica o filtro também ao cache, para manter explícita a população analisada.
df_ideb = df_ideb.loc[
    df_ideb["SG_UF"].eq("SP") &
    df_ideb["REDE"].isin(["Municipal", "Estadual", "Federal"])
].copy()

colunas_resultado = ["VL_OBSERVADO_2023", "VL_OBSERVADO_2025"]
for coluna in colunas_resultado:
    df_ideb[coluna] = pd.to_numeric(df_ideb[coluna], errors="coerce")

if df_ideb["ID_ESCOLA"].isna().any() or df_ideb["ID_ESCOLA"].duplicated().any():
    raise ValueError("ID_ESCOLA precisa ser único e não nulo na base de IDEB filtrada.")

df_ideb = df_ideb.rename(
    columns={
        "ID_ESCOLA": "CO_ENTIDADE",
        "CO_MUNICIPIO": "CO_MUNICIPIO_IDEB"
    }
)

# Left join preserva toda a coorte; escolas sem nota disponível ficam com valor ausente.
df_painel_escolas_com_ideb = df_painel_escolas.merge(
    df_ideb,
    on="CO_ENTIDADE",
    how="left",
    validate="one_to_one"
)

print("Escolas na coorte pareada:", len(df_painel_escolas_com_ideb))
print("Escolas com registro de IDEB:", df_painel_escolas_com_ideb["REDE"].notna().sum())
print("Escolas com IDEB numérico em 2023:", df_painel_escolas_com_ideb["VL_OBSERVADO_2023"].notna().sum())
print("Escolas com IDEB numérico em 2025:", df_painel_escolas_com_ideb["VL_OBSERVADO_2025"].notna().sum())
print("Escolas com IDEB numérico nos dois anos:", df_painel_escolas_com_ideb[colunas_resultado].notna().all(axis=1).sum())

caminho_painel_ideb = DADOS_TRATADOS / "painel_escolas_sp_2023_2025_com_ideb.csv"
df_painel_escolas_com_ideb.to_csv(
    caminho_painel_ideb,
    index=False,
    encoding="utf-8-sig"
)
print("Painel com IDEB salvo em:", caminho_painel_ideb.relative_to(PASTA_PROJETO))
```
